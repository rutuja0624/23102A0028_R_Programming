```r

# ============================================================
# Air Quality Data Cleaning Assignment (R)
# Dataset: Beijing Multi-Site Air Quality Dataset
# Station: Aotizhongxin
# ============================================================

# ---------------------------
# Install (run once if needed)
# ---------------------------
if(!require(ggplot2)) install.packages("ggplot2", repos="https://cloud.r-project.org")
library(ggplot2)

# ---------------------------
# Task 1 : Import Dataset
# ---------------------------
tryCatch({

  unzip("PRSA2017_Data_20130301-20170228.zip")

  air_data <- read.csv(
    "PRSA_Data_20130301-20170228/PRSA_Data_Aotizhongxin_20130301-20170228.csv",
    stringsAsFactors = FALSE
  )

  cat("Dataset loaded successfully.\n")

}, error=function(e){
  stop(paste("Error:",conditionMessage(e)))
})

cat("\nFirst Six Records\n")
print(head(air_data))

cat("\nStructure\n")
str(air_data)

cat("\nRows:",nrow(air_data))
cat("\nColumns:",ncol(air_data))

cat("\nContains Missing Values:",any(is.na(air_data)))
cat("\nTotal Missing Values:",sum(is.na(air_data)))

# ---------------------------
# Task 2 : NA NULL NaN
# ---------------------------
temperature <- c(28,30,NA,32)
missing_object <- NULL
undefined_value <- 0/0

cat("\n\nNA Example\n")
print(is.na(temperature))

cat("\nNULL Example\n")
print(is.null(missing_object))

cat("\nNaN Example\n")
print(is.nan(undefined_value))

# ---------------------------
# Task 3 : Missing Summary
# ---------------------------
missing_summary <- function(df){

  variables <- c("PM2.5","PM10","SO2",
                 "NO2","TEMP","WSPM","wd")

  result <- data.frame()

  for(v in variables){

    total <- nrow(df)
    missing <- sum(is.na(df[[v]]))
    percent <- round((missing/total)*100,2)

    if(percent>20){
      warning(paste(v,"contains more than 20% missing values"))
    }

    result <- rbind(result,
                    data.frame(
                      Variable=v,
                      Total_Records=total,
                      Missing_Values=missing,
                      Missing_Percentage=percent
                    ))
  }

  return(result)
}

summary_table <- missing_summary(air_data)

cat("\nMissing Summary\n")
print(summary_table)

# ---------------------------
# Task 4 : Pollution Ratio
# ---------------------------
air_data$pollution_ratio <- air_data$PM2.5/air_data$PM10

cat("\nNA:",sum(is.na(air_data$pollution_ratio)))
cat("\nNaN:",sum(is.nan(air_data$pollution_ratio)))
cat("\nInfinite:",sum(is.infinite(air_data$pollution_ratio)))

air_data$pollution_ratio[
  is.nan(air_data$pollution_ratio) |
  is.infinite(air_data$pollution_ratio)
] <- NA

# ---------------------------
# Task 5 : Median Replacement
# ---------------------------
numeric_variables <- c("PM2.5","PM10","SO2",
                       "NO2","TEMP","WSPM")

before <- c()
after <- c()

for(v in numeric_variables){

  if(v %in% names(air_data)){

    missing_before <- sum(is.na(air_data[[v]]))

    med <- median(air_data[[v]],na.rm=TRUE)

    air_data[[v]][is.na(air_data[[v]])] <- med

    missing_after <- sum(is.na(air_data[[v]]))

    before <- c(before,missing_before)
    after <- c(after,missing_after)

    cat("\n------------------------")
    cat("\nVariable:",v)
    cat("\nMissing Before:",missing_before)
    cat("\nMedian:",med)
    cat("\nMissing After:",missing_after)
    cat("\n------------------------\n")
  }
}

# ---------------------------
# Task 6 : Mode Function
# ---------------------------
calculate_mode <- function(x){

  x <- na.omit(x)

  uniq <- unique(x)

  uniq[which.max(tabulate(match(x,uniq)))]
}

before_wd <- sum(is.na(air_data$wd))

mode_wd <- calculate_mode(air_data$wd)

air_data$wd[is.na(air_data$wd)] <- mode_wd

after_wd <- sum(is.na(air_data$wd))

cat("\nMode of wd:",mode_wd)

# ---------------------------
# Task 7 : Error Handling
# ---------------------------
clean_variable <- function(df,var){

  tryCatch({

    if(!(var %in% names(df)))
      stop("Variable does not exist.")

    if(!is.numeric(df[[var]]))
      stop("Variable is categorical.")

    if(all(is.na(df[[var]])))
      stop("Variable contains only missing values.")

    med <- median(df[[var]],na.rm=TRUE)

    if(is.na(med))
      stop("Median cannot be calculated.")

    df[[var]][is.na(df[[var]])] <- med

    return(df[[var]])

  },error=function(e){

    message(conditionMessage(e))

  })
}

clean_variable(air_data,"PM2.5")
clean_variable(air_data,"wd")
clean_variable(air_data,"XYZ")

# ---------------------------
# Task 8 : Comparison Table
# ---------------------------
comparison <- data.frame(
  Variable=c(numeric_variables,"wd"),
  Missing_Before=c(before,before_wd),
  Missing_After=c(after,after_wd)
)

comparison$Values_Replaced <-
comparison$Missing_Before-comparison$Missing_After

cat("\nComparison Table\n")
print(comparison)

# ---------------------------
# Task 9 : Visualization
# ---------------------------
plot_data <- data.frame(
Variable=rep(comparison$Variable,2),
Missing=c(comparison$Missing_Before,
          comparison$Missing_After),
Status=rep(c("Before","After"),
each=nrow(comparison))
)

ggplot(plot_data,
aes(Variable,Missing,fill=Status))+
geom_bar(stat="identity",
position="dodge")+
labs(title="Missing Values Before and After Cleaning",
x="Variables",
y="Missing Values")+
theme_minimal()

# ---------------------------
# Task 10 : Export
# ---------------------------
write.csv(
air_data,
"cleaned_air_quality_data.csv",
row.names=FALSE
)

cat("\nCleaned dataset exported successfully.\n")

cat("\nInterpretation:\n")
cat("The dataset contained missing values in numerical and categorical variables. Numerical values were replaced using the median, while missing wind direction values were replaced with the mode. Invalid values in the pollution ratio were converted to NA before cleaning. The comparison table and visualization show that missing values were effectively handled. The cleaned dataset is now suitable for further statistical analysis and machine learning.\n")
```
