```r
x <- c(11, NA, 13, NA)
print(x)
print(is.na(x))
print(mean(x))
print(mean(x, na.rm = TRUE))
print(is.null(NULL))
print(is.na(NA))
```

```r
x <- c(11, NA, 13, NA)
print(which(is.na(x)))
print(sum(is.na(x)))
print(complete.cases(x))
y <- na.omit(x)
print(y)
print(mean(y))
```
