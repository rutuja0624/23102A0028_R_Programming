```r
x <- c(2, 4, 6, 8, 10, 12)
count_values <- function(x) {
  count <- 0
  for (xval in x) {
    if (xval / 2 > 3) count <- count + 1
  }
  count
}
print(count_values(x))

child <- c("child1", "child2", "child3")
sweet <- c("sweet1", "sweet2", "sweet3")
for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

```r
drink <- c("coffee", "lemonade", "tea", "juice")
for (x in drink) {
  if (x == "tea") break
  print(x)
}

for (x in drink) {
  if (x == "lemonade") next
  print(x)
}
```

```r
i <- 1
while (i < 10) {
  print(i^2)
  i <- i + 2
}
```

```r
i <- 1
repeat {
  print(i^2)
  i <- i + 2
  if (i > 10) break
}
```

```r
square_value <- function(x) {
  result <- x^2
  return(result)
}
print(square_value(5))

sum_two <- function(a, b) {
  a + b
}
print(sum_two(10, 20))

greet <- function(name = "Student") {
  paste("Hello", name)
}
print(greet())
print(greet("Rutuja"))
```
