```r
A <- matrix(1:12, nrow = 4, ncol = 3)
rownames(A) <- c("r1", "r2", "r3", "r4")
colnames(A) <- c("c1", "c2", "c3")
print(A)
constant_matrix <- matrix(2, nrow = 4, ncol = 2)
print(constant_matrix)
```

```r
I <- diag(1, nrow = 3, ncol = 3)
D <- diag(5, nrow = 3, ncol = 3)
print(I); print(D)
X <- matrix(1:8, nrow = 4, ncol = 2, byrow = TRUE)
print(X)
print(t(X))
print(rowSums(X))
print(colSums(X))
print(rowMeans(X))
print(colMeans(X))
```

```r
X <- matrix(1:15, nrow = 5, ncol = 3, byrow = TRUE)
print(X)
print(X[3, ])
print(X[, 2])
print(X[4:5, 2:3])
print(X[c(1, 4), c(1, 3)])
```

```r
A <- matrix(1:6, nrow = 2, byrow = TRUE)
B <- matrix(6:1, nrow = 2, byrow = TRUE)
print(A + 5)
print(A - 5)
print(A * 2)
print(A / 2)
print(A + B)
print(A - B)
print(A * B)
print(A %*% t(B))
```

```r
A <- matrix(c(4, 7, 2, 6), nrow = 2, byrow = TRUE)
B <- matrix(c(1, 2), nrow = 1)
print(rbind(A, B))
print(cbind(A, c(9, 8)))
print(solve(A))
print(eigen(A))
print(crossprod(A))
```

```r
x <- c(1, 2, 3, 4, 5)
print(x > 3)
print(x == 3)
print(x != 3)
print((x > 1) & (x < 5))
print((x < 2) | (x > 4))
print(! (x == 3))
```
